```python
import pandas as pd
import re
import nltk

# Download NLTK datasets for stopwords and tokenization
nltk.download('punkt')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Set of English stop words and stemmer
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
print("Libraries imported successfully!")
```

```python
# Load the CSV dataset
df = pd.read_csv('twitter_training.csv', header=None)
df.columns = ['ID', 'Entity', 'Sentiment', 'Text']

print("Dataset Shape:", df.shape)
df.head()
```

```python
# Check missing values
print("Missing values per column:\n", df.isnull().sum())

# Drop rows where text is missing
df = df.dropna(subset=['Text']).reset_index(drop=True)
print("Dataset Shape after dropping missing text:", df.shape)
```

```python
def preprocess_text(text):
    # 1. Lowercasing
    text = str(text).lower()

    # 2. Remove URLs, mentions, numbers, and special characters/punctuation
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)  # remove URLs
    text = re.sub(r'[^a-zA-Z\s]', '', text)               # remove special characters & numbers

    # 3. Tokenization
    tokens = text.split()

    # 4. Stop Words Removal & 5. Stemming
    cleaned_tokens = [stemmer.stem(word) for word in tokens if word not in stop_words]

    # Rejoin tokens into a single clean string
    return ' '.join(cleaned_tokens)
```

```python
# Apply the preprocessing function to the 'Text' column
df['Cleaned_Text'] = df['Text'].apply(preprocess_text)

# Display original text vs cleaned text
df[['Text', 'Cleaned_Text']].head(10)
```

```python
# Export cleaned data to a new CSV file
df.to_csv('cleaned_twitter_data.csv', index=False)
print("Preprocessed data saved to cleaned_twitter_data.csv!")
```
