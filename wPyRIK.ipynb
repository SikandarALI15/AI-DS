1. Create a class 'BankAccount' with attributes account holder name, account number, and
balance. Add methods for deposit and withdrawal. Implement logic to prevent withdrawal if balance
is insufficient.

```python
class BankAccount:
    def __init__(self, holder_name, account_number, balance=0):
        self.holder_name = holder_name
        self.account_number = account_number
        self.balance = balance

    def deposit(self):
        amount = int(input("Enter amount: "))
        self.balance += amount
        print("Deposit successful")

    def withdrawal(self):
        amount = int(input("Enter amount: "))
        if amount > self.balance:
            print("Insufficient balance!")
        else:
            self.balance -= amount
            print("Withdrawal successful")

ac1 = BankAccount("Ali", "12345", 1000)

ac1.balance
ac1.withdrawal( )
```

```python
class Student: 
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks

    def calculate_grade(self):
        if self.marks > 80 and self.marks <101:
            print("A")

        elif self.marks > 70 and self.marks <80:
            print("B")
            
        elif self.marks > 60 and self.marks <70:
            print("C")    

        else:
            print("Fail")

st1=Student("Ali",88) 
st1.calculate_grade()
```

6. Design an abstract class 'Shape' with a method 'area()'. Create subclasses 'Circle' and
'Rectangle' and implement area calculation logic.

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

c = Circle(5)
r = Rectangle(4, 6)

print("Circle Area:", c.area())
print("Rectangle Area:", r.area())
```

7. Create a class 'HospitalPatient' with attributes name, age, and disease. Add a method to
determine priority (Emergency if age > 60 or serious disease).

```python
class HospitalPatient:
    def __init__(self, name, age, disease):
        self.name = name
        self.age = age
        self.disease = disease

    def priority(self):
        if self.age > 60 or self.disease == "Serious":
            return "Emergency"
        return "Normal"


p1 = HospitalPatient("Ali", 65, "Fever")
p2 = HospitalPatient("Sara", 25, "Serious")

print(p1.priority())
print(p2.priority())
```

8. Build a class 'ShoppingCart' where user can add items with price. Implement logic to calculate
total bill and apply discount if total exceeds a certain amount.

```python
class ShoppingCart:
    def __init__(self):
        self.total = 0

    def add_item(self, price):
        self.total += price

    def bill(self):
        if self.total > 1000:
            return self.total * 0.9   # 10% discount
        return self.total


cart = ShoppingCart()

cart.add_item(500)
cart.add_item(700)

print("Total Bill:", cart.bill())
```

10. Design a class 'Library' where users can borrow and return books. Add logic to check
availability before issuing a book.

```python
class Library:
    def __init__(self):
        self.books = ["Python", "Java", "C++"]

    def borrow_book(self, book):
        if book in self.books:
            self.books.remove(book)
            print(book, "issued successfully")
        else:
            print(book, "is not available")

    def return_book(self, book):
        self.books.append(book)
        print(book, "returned successfully")


lib = Library()

lib.borrow_book("Python")
lib.borrow_book("Python") 

lib.return_book("Python")
```
