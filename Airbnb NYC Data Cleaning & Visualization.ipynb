```python
data["last_review"]=data["last_review"].fillna("No review")
# Yhan last review ko "No review" se fill isliye kiya because data se decision liya jata hai to ye fake data dene se better hai
```

```python
data["reviews_per_month"]=data["reviews_per_month"].fillna(data.mode)
```

```python
data["calculated_host_listings_count"]=data["calculated_host_listings_count"].fillna(data["calculated_host_listings_count"].mode()[0])
```

```python
data["latitude"]=data["latitude"].fillna(data["latitude"].mode()[0])
```

```python
data.drop_duplicates(inplace=True)
```

```python
data["price"] = data["price"].str.replace("$", "")
data["price"] = data["price"].str.replace(",", "")
data["price"] = pd.to_numeric(data["price"], errors="coerce")
```

```python
data.price.dtype
```

```python
data["price"]=data["price"].fillna(data["price"].median())
```

```python
data["price"]=data["price"].astype(int)
```

```python
Q1 = data["price"].quantile(0.25)
Q3 = data["price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

price_outliers = data[(data["price"] < lower) | (data["price"] > upper)]
```

```python
data['price'] = np.where(data['price'] > upper, upper, data['price'])
data['price'] = np.where(data['price'] < lower, lower, data['price'])

```
