```python
data=data.drop("No",axis=1)
```

```python
data.columns
```

```python
x=data.drop('Y house price of unit area',axis=1)
```

```python
y=data['Y house price of unit area']
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
comparrision=pd.DataFrame({
    "Actual":y_test.values,
    "Predicted":y_pred
})
```

```python
comparrision
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mae=mean_absolute_error(y_test,y_pred)
```

```python
mae
```

```python
from sklearn.metrics import r2_score
```

```python
r2=r2_score(y_test,y_pred)
```

```python
r2
```

```python
def pred_price(): 
    transaction_date=float(input("Enter  transaction date: "))
    house_age=float(input("Enter house age: "))
    mrt_distance=float(input("Enter Metro distance: "))
    convenience_store =int(input("Enter convenience store: "))
    latitued=float(input("Enter latitued: "))
    longitued=float(input("Enter longitued: "))

    prediction=lr.predict([[
        transaction_date, 
        house_age, 
        mrt_distance, 
        convenience_store, 
        latitued, 
        longitued
    ]])
    return prediction[0]
```
